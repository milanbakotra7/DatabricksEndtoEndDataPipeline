###  Dynamic Capibilities

```python
dbutils.widgets.text("filename", "")

```

```python
p_file_name = dbutils.widgets.get("filename")
```

## Data Reading 

```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "parquet")\
    .option("cloudFiles.schemaLocation", f"abfss://bronze@strdatabricks.dfs.core.windows.net/checkpoint_{p_file_name}")\
    .load(f"abfss://source@strdatabricks.dfs.core.windows.net/{p_file_name}")
```

## Data Writing into Bronze Layer

```python
df.writeStream.format("parquet")\
    .outputMode("append")\
    .option("checkpointLocation", f"abfss://bronze@strdatabricks.dfs.core.windows.net/checkpoint_{p_file_name}")\
    .option("path", f"abfss://bronze@strdatabricks.dfs.core.windows.net/{p_file_name}")\
    .trigger(once=True)\
    .start()
```
