```python
%skip
catalog_name = "datapipeline"
schema_names = ["bronze", "silver", "gold"]

table_list = []

# Step 1: Get all tables
for schema in schema_names:

    tables_df = spark.sql(
        f"SHOW TABLES IN {catalog_name}.{schema}"
    )

    for row in tables_df.collect():
        table_list.append((schema, row.tableName))


# Step 2: Drop all tables
for schema, table in table_list:

    full_table_name = f"{catalog_name}.{schema}.{table}"

    spark.sql(f"DROP TABLE IF EXISTS {full_table_name}")

    print(f"Dropped: {full_table_name}")

```

```python
catalog_name = "datapipeline"
schema_names = ["bronze", "silver", "gold"]

table_list = []

for schema in schema_names:

    tables_df = spark.sql(
        f"SHOW TABLES IN {catalog_name}.{schema}"
    )

    for row in tables_df.collect():
        table_list.append(
            (schema, row.tableName)
        )

print(table_list)
```

```python
results = []

for schema, table in table_list:

    full_table_name = f"{catalog_name}.{schema}.{table}"

    count = spark.table(full_table_name).count()

    results.append(
        (schema, table, count)
    )
```

```python
count_df = spark.createDataFrame(
    results,
    ["schema_name", "table_name", "record_count"]
)

display(count_df)
```
